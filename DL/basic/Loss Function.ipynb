```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1/1+np.exp(-x)
```

```python
def relu(x):
    return np.maximum(0,x)
```

```python
x=np.linspace(-50,50,50)
y1=sigmoid(x)
y2=relu(x)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.axhline(0, color='black', lw=1) # 画出 x 轴
ax1.axvline(0, color='black', lw=1) # 画出 y 轴
ax1.plot(x,y1)
ax2.axhline(0, color='black', lw=1) # 画出 x 轴
ax2.axvline(0, color='black', lw=1) # 画出 y 轴
ax2.plot(x,y2)
```
