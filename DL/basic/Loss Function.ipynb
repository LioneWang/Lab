```python
import torch
import torch.nn as nn
import numpy as np
import torch.nn.functional as F
```

```python
def cross_entropy(logits,target):
    # (batch_size,num_classes)->(batch_size,1)->(batch_size)
    max_logits,_=torch.max(logits,dim=-1,keepdim=True)
    # (batch_size,num_classes)
    logits_stable=logits-max_logits
    # sum会降低维度
    # (batch_size)
    log_sum_exp=torch.log(torch.sum(torch.exp(logits_stable),dim=-1))
    # (batch_size)
    target_logits=logits_stable.gather(1,target.unsqueeze(1)).squeeze(1)
    loss=-target_logits+log_sum_exp
    return loss.mean()
```

# 分类任务loss

```python
num_classes=5
batch_size=3
logits=torch.randn(batch_size,num_classes)
target=torch.tensor([1,0,4])
loss_official=F.cross_entropy(logits,target)
loss_custom=cross_entropy(logits,target)
print(f"官方 Loss: {loss_official.item():.6f}")
print(f"手搓 Loss: {loss_custom.item():.6f}")
print(f"误差: {torch.abs(loss_official - loss_custom).item():.1e}")
```

# 自回归任务loss

```python
batch_size=5
seq_len=10
vocab_size=1000
logits=torch.randn(batch_size,seq_len,vocab_size)
#targets是【0，vocab】之间的随机整数
targets=torch.randint(0,vocab_size,(batch_size,seq_len))
logits=logits.view(-1,logits.shape[-1])
targets=targets.view(-1)
loss=F.cross_entropy(logits,targets)
print(f"Loss value: {loss.item():.4f}")
```
