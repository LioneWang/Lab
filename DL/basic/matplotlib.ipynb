# linspace

```python
# 在一个区间范围内生成
x = np.linspace(start=0, stop=2*np.pi, num=400)
y = np.sin(x**2)
```

# figure&subplot

```python
#创建画板
fig=plt.figure()
ax = fig.subplots()
ax.plot(x, y)
ax.set_title('Simple plot')
```

```python
fig=plt.figure()
ax1, ax2 = fig.subplots(nrows=1, ncols=2, sharey=True)
ax1.plot(x,y)
ax1.set_title('Sharing Y axis')
ax2.scatter(x, y)
```

```python
fig=plt.figure()
axes = fig.subplots(2, 2, subplot_kw=dict(projection='polar'))
print(axes.shape)
axes[0, 0].plot(x, y)
axes[1, 1].scatter(x, y)
```
